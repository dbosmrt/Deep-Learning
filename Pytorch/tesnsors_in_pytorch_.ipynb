#Tensor Shapes

```python
x = torch.tensor([[1,2,3],[4,5,6]])
```

```python
x.shape
```

```python
torch.empty_like(x)
```

```python
torch.zeros_like(x)
```

```python
torch.ones_like(x)
```

```python
torch.rand_like(x, dtype=torch.float32)
"""
Reason:
x has integers in the matrix where as rand generates floats between 0 and 1.
have to explictly tell the data type.
"""
```
